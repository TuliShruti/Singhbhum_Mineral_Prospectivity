## Interpretation checklist
- Identify whether anomalous stations form spatial clusters or isolated points.
- Compare anomaly scores with nearby documented copper occurrences, while accounting for distance.
- Do not call high anomaly score a high copper probability.
- Do not extrapolate outside the observed station footprint.
- Discuss the small number of occurrence points and the absence of a verified basin polygon/geochemical layer.
